```python
# Transaction data
transactions = [
    ["BREAD", "MILK"],
    ["BREAD", "DIAPER", "BEER", "EGGS"],
    ["MILK", "DIAPER", "BEER", "COKE"],
    ["BREAD", "MILK", "DIAPER", "BEER"],
    ["BREAD", "MILK", "DIAPER", "COKE"]
]

min_support = 3   # itemset must occur at least 3 times

# Find all items
items = []

for t in transactions:
    for item in t:
        if item not in items:
            items.append(item)

# Find frequent 1-itemsets
print("Frequent 1-itemsets:")

frequent = []

for item in items:
    count = 0

    for t in transactions:
        if item in t:
            count += 1

    if count >= min_support:
        frequent.append([item])
        print(item, "=", count)

# Find frequent 2-itemsets
print("\nFrequent 2-itemsets:")

for i in range(len(frequent)):
    for j in range(i + 1, len(frequent)):

        item1 = frequent[i][0]
        item2 = frequent[j][0]

        count = 0

        for t in transactions:
            if item1 in t and item2 in t:
                count += 1

        if count >= min_support:
            print("{" + item1 + ", " + item2 + "}", "=", count)
```
