```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

#Z - Score Normalization
x = ([10,20,30,40,50])
mean = sum(x)/len(x)
variance = sum((xi - mean)**2 for xi in x)/len(x)
std_dev = variance**0.5
z = [(xi - mean)/std_dev for xi in x]
print("\nZ-Sdcore Normalization values are :",z)

#min-max normalization
x = ([10,20,30,40,50])
min_val = min(x)
max_val = max(x)
normalized = [(xi - min_val)/(max_val - min_val) for xi in x]
print("\nMin-Max Normalization values are:",normalized)

#decimal scaling
import math
x = ([10,20,30,40,50])
i = math.floor(math.log10(max(x)))
normalized = [xi/(10**i) for xi in x]
print("\nDecimal Scaling values are:",normalized)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('/content/Titanic-Dataset.csv')

df = df.drop('Name', axis=1)
display(df.head())
```

```python
import math
max_fare_abs = df['Fare'].abs().max()

if max_fare_abs > 0:
    j = math.floor(math.log10(max_fare_abs))
else:
    j = 0
df['Fare_Decimal_Scaled'] = df['Fare'] / (10**j)

display(df[['Fare', 'Fare_Decimal_Scaled']].head(10))
```

```python
min_age = df['Age'].min()
max_age = df['Age'].max()
df['Age_MinMax_Scaled'] = (df['Age'] - min_age) / (max_age - min_age)

min_fare = df['Fare'].min()
max_fare = df['Fare'].max()
df['Fare_MinMax_Scaled'] = (df['Fare'] - min_fare) / (max_fare - min_fare)

display(df[['Age', 'Age_MinMax_Scaled', 'Fare', 'Fare_MinMax_Scaled']].head(10))
```

```python
mean_age = df['Age'].mean()
std_age = df['Age'].std()
df['Age_ZScore_Scaled'] = (df['Age'] - mean_age) / std_age

mean_fare = df['Fare'].mean()
std_fare = df['Fare'].std()
df['Fare_ZScore_Scaled'] = (df['Fare'] - mean_fare) / std_fare

display(df[['Age', 'Age_ZScore_Scaled', 'Fare', 'Fare_ZScore_Scaled']].head(10))
```
