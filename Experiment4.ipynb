```python
import matplotlib.pyplot as plt

fruits = ['Apples', 'Bananas', 'Cherries', 'Dates']
sales = [400, 350, 300, 450]

plt.bar(fruits, sales)
plt.title('Fruit Sales')
plt.xlabel('Fruits')
plt.ylabel('Sales')
plt.show()
```

```python
import matplotlib.pyplot as plot

x = [13,12,23,64,85,96]
y = [1,2,3,4,5,6]

plot.figure(figsize=(8, 6))
plot.plot(x, y)
plot.title('Simple Line Plot')
plot.xlabel('X-axis')
plot.ylabel('Y-axis')
plot.grid(True)
plot.show()
```

```python
import matplotlib.pyplot as plt

a = [1, 2, 3, 4, 5]
b = [1, 4, 6, 8, 10]

plt.boxplot([a, b], showfliers=False)
plt.xlabel('Data Set')
plt.ylabel('Values')
plt.title('Simple Box Plot')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt


x = [12, 45, 7, 32, 89, 54, 23, 67, 14, 91]
y = [99, 31, 72, 56, 19, 88, 43, 61, 35, 77]
plt.scatter(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Scatter Plot')
plt.grid()
plt.show()
```

```python
import matplotlib.pyplot as plt

a = ["a", "b", "c", "d", "e" ]
b = [1, 4, 6, 8, 10]

plt.pie(b, labels=a, autopct='%1.1f%%')

plt.title('Simple Pie Chart')
plt.legend(loc='upper right')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
data = [12, 15, 15, 18, 20, 21, 21, 21, 22, 25, 28, 30, 35, 40]
plt.hist(data, bins=10, color='skyblue', edgecolor='black')
plt.title('Simple Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(axis='y')
plt.show()
```
